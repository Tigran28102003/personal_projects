## Question 1

Build a simple single layer neural network to classify AND/OR logic gates using different (step, sigmoid, and ReLU, etc.) activation functions (five minimum). Train each model and explain why some activations succeed while others fail for these linearly separable problems. Compare the results and discuss the effect of activation functions and your suggestion about choosing activation function of different tasks.

```python
# generate data for logic gates

def make_logic_data(gate="AND"):

    X = np.array([[0,0],[0,1],[1,0],[1,1]], dtype=float)

    if gate == "AND":
        y = np.array([0,0,0,1], dtype=float)
    elif gate == "OR":
        y = np.array([0,1,1,1], dtype=float)
    elif gate == "XOR":
        y = np.array([0,1,1,0], dtype=float)
    else:
        raise ValueError("unknown gate")

    return X, y
```

```python
# activation functions
def tanh(z):
    return np.tanh(z)

def relu(z):
    return np.maximum(0, z)

def leaky_relu(z, a=0.01):
    return np.where(z > 0, z, a * z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# & accuracy metric
def accuracy(y_true, y_prob, thr=0.5):
    y_pred = (y_prob >= thr).astype(float)
    return (y_pred == y_true).mean()
```

```python
def train_single_layer(X, y, act_name, lr=0.1, epochs=2000):
    """
    Trains a single layer neural network with specified activation function.
    Returns predicted probabilities for the training data.
    """
    w = np.random.normal(0, 0.5, size=(X.shape[1],))
    b = 0.0


    for _ in range(epochs):
        z = X @ w + b

        if act_name == "step":
            # couldn't be trained with GD, but can be with Perceptron rule
            y_hat = (z >= 0).astype(float)
            err = y - y_hat
            w += lr * (X.T @ err)
            b += lr * err.sum()

        else:
            # just GD for the rest
            if act_name == "sigmoid":
                a = sigmoid(z)
                grad_z = (a - y)

            elif act_name == "tanh":
                a = tanh(z)
                da_dz = 1 - a**2
                grad_z = (a - y) * da_dz

            elif act_name == "relu":
                a = relu(z)
                da_dz = (z > 0).astype(float)
                grad_z = (a - y) * da_dz

            elif act_name == "leaky_relu":
                a = leaky_relu(z)
                da_dz = np.where(z > 0, 1.0, 0.01)
                grad_z = (a - y) * da_dz

            else:
                raise ValueError("unknown activation")

            w -= lr * (X.T @ grad_z) / len(X)
            b -= lr * grad_z.mean()


    z = X @ w + b
    if act_name == "step":
        y_prob = (z >= 0).astype(float)
    elif act_name == "sigmoid":
        y_prob = sigmoid(z)
    elif act_name == "tanh":
        y_prob = (tanh(z) + 1) / 2
    else:
        y_prob = sigmoid(z)

    return y_prob

acts = ["step", "sigmoid", "tanh", "relu", "leaky_relu"]

for gate in ["AND", "OR"]:
    X, y = make_logic_data(gate)

    print(f"\nGATE = {gate}")

    for act_name in acts:
        lr = 0.2 if act_name == "step" else 0.1
        y_prob = train_single_layer(X, y, act_name, lr=lr, epochs=1500)  # fixed epochs
        print(f"  {act_name:10s}  acc={accuracy(y, y_prob):.2f}  probs={np.round(y_prob, 3)}")
```

AND and OR are linearly separable, so a single neuron *can* solve them. In your runs (1500 epochs), the difference comes from how well the activation matches the learning rule and the “probability” interpretation.

The **step (perceptron)** model reaches perfect accuracy for both gates because it does not need gradients at all. It simply updates weights based on classification mistakes, and for linearly separable data that is enough to find a separating boundary.

The **sigmoid** model also gets perfect accuracy on both AND and OR. This is the most “natural” setup for binary classification: the output is already in ([0,1]), and with a cross-entropy style gradient the updates stay stable. Your probabilities reflect that: negatives are close to 0 and positives are pushed toward 1 (for AND, the positive case goes to ~0.84; for OR, most positives are ~0.95–1.00).

With **tanh**, performance drops (AND 0.50, OR 0.75). The main reason is a mismatch between the activation range and the labels: tanh outputs in ([-1,1]), while the targets are (0/1). With an MSE-like training signal, the model often settles into a “compromise” where negative points sit around 0.4–0.6 instead of near 0, so thresholding makes mistakes. That is exactly what you see for AND: some inputs that should be class 0 end up around 0.62.

**ReLU** and **leaky ReLU** behave similarly here (AND 0.50, OR 0.75). These activations are great in hidden layers, but as a final activation for binary classification they are awkward: the output is not a probability, and after your mapping it tends to land near the decision boundary (you even get exact 0.5 values for AND), which makes predictions unstable. Leaky ReLU fixes the “zero gradient” issue for negative inputs, but it does not fix the bigger problem: this is still not a well-calibrated output activation for a 0/1 classifier in this simple setup.

In practice, for a single-layer binary classifier the most reliable choice is **sigmoid in the output with a cross-entropy loss**. If you want to use ReLU or leaky ReLU, it is usually better to put them in **hidden layers**, and keep the last layer sigmoid if you interpret the output as a probability.
